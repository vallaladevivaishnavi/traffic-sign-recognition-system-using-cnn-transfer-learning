# 1.2 Introduction to CNN

CNNs have become the foundation of many modern artificial intelligence applications, including image classification, object detection, facial recognition, medical image analysis, autonomous driving, and security surveillance. Their ability to automatically learn hierarchical features from raw image data has made them one of the most successful deep learning models.

# 3.3 Image Resizing (30 × 30)

Image resizing is a crucial preprocessing step in deep learning-based image classification projects. The images in the GTSRB dataset have different resolutions and dimensions, which cannot be directly provided as input to a Convolutional Neural Network (CNN). Therefore, all images were resized to a uniform dimension of 30 × 30 pixels before training the model. This standardization ensures consistency across the dataset and allows the CNN to process every image with the same input shape. Resizing also reduces computational complexity, decreases memory usage, and speeds up the training process while preserving the essential visual features of traffic signs such as shapes, colors, and symbols. By converting all images to a fixed size of 30 × 30 pixels, the model can efficiently learn meaningful patterns and achieve better classification performance.


# 3.4 Image Normalization

Image normalization is an important preprocessing technique used to improve the performance and stability of deep learning models. In the GTSRB dataset, pixel values range from 0 to 255, where 0 represents black and 255 represents white. Directly feeding these large pixel values into the CNN can slow down the learning process and negatively affect model convergence. Therefore, all image pixel values were normalized by dividing them by 255, transforming the range from 0–255 to 0–1. This scaling process helps the neural network learn more efficiently by ensuring that all input features have a similar range. Normalization reduces numerical instability, accelerates gradient descent optimization, and improves the overall accuracy and training speed of the CNN model. As a result, the model can learn meaningful traffic sign features more effectively and achieve better classification performance.


#3.5 Data Augmentation

Data augmentation is a powerful preprocessing technique used to increase the diversity of training data without collecting additional images. In the GTSRB dataset, traffic sign images may appear under different real-world conditions such as varying lighting, rotations, zoom levels, camera angles, and positional shifts. To make the Convolutional Neural Network (CNN) more robust and capable of handling these variations, data augmentation techniques were applied to the training images. Operations such as rotation, width shifting, height shifting, zooming, shearing, and brightness adjustments were performed to generate modified versions of existing images while preserving their original class labels. This process helps reduce overfitting, improves the model’s ability to generalize to unseen data, and enhances overall classification accuracy. By exposing the CNN to a wider variety of image conditions during training, data augmentation enables the model to learn more meaningful and invariant features of traffic signs.


# 3.6 Label Encoding

Label encoding is an essential preprocessing step used to convert categorical class labels into a numerical format that can be understood by the Convolutional Neural Network (CNN). In the GTSRB dataset, each traffic sign category is represented by a unique class identifier ranging from 0 to 42, corresponding to 43 different traffic sign classes. Since neural networks cannot directly process categorical labels, these class labels were transformed into a machine-readable format using one-hot encoding. In one-hot encoding, each class is represented as a binary vector where only one element is assigned a value of 1, while all other elements are assigned a value of 0. This representation prevents the model from assuming any ordinal relationship between classes and allows it to perform multi-class classification effectively. Label encoding ensures that the CNN can accurately learn the relationship between input images and their corresponding traffic sign categories.

```python
data = []
labels = []
classes = 43

# CORRECT PATH AFTER EXTRACTION
base_path = "/content/GTSRB"

for i in range(classes):

    path = os.path.join(base_path, 'train', str(i))  #Creates path to current class folder.

    images = os.listdir(path)

    for a in images:

        try:
            img_path = os.path.join(path, a)
            img = cv2.imread(img_path)

            img = cv2.resize(img, (30, 30))

            data.append(img)

            labels.append(i)

        except Exception as e:
            print("Error loading image:", a)
            print(e)

data = np.array(data)
labels = np.array(labels)

print("Data Shape:", data.shape)
print("Labels Shape:", labels.shape)
```

3.7 Train-Test Split

Train-test splitting is an important step in machine learning and deep learning projects because it helps evaluate how well a model performs on unseen data. After preprocessing the GTSRB dataset, the images and their corresponding labels were divided into two subsets: a training set and a testing set. The training set was used to train the Convolutional Neural Network (CNN) and learn the patterns present in different traffic sign categories, while the testing set was used to evaluate the model's performance on new images that were not seen during training. In this project, the dataset was split using the train_test_split() function from Scikit-learn, ensuring that both training and testing datasets contained representative samples from all 43 traffic sign classes. This separation helps prevent overfitting and provides a realistic measure of the model’s generalization capability. A properly performed train-test split is essential for obtaining reliable and unbiased performance metrics such as accuracy, precision, recall, and F1-score.

```python
X_train, X_test, y_train, y_test = train_test_split(data,labels,test_size=0.2,random_state=42, stratify=labels
)

print("X_train:", X_train.shape)

print("X_test:",X_test.shape)

print("y_train:",y_train.shape)

print("y_test:", y_test.shape)
```

# 4. Convolutional Neural Network (CNN)

4.1 Introduction to CNN (Convolutional Neural Network)

A Convolutional Neural Network (CNN) is a specialized type of Deep Learning model designed for processing and analyzing image data. CNNs are highly effective in image classification tasks because they can automatically learn important visual features such as edges, shapes, textures, colors, and patterns directly from images without requiring manual feature extraction. Unlike traditional machine learning algorithms, CNNs use multiple layers, including convolutional layers, pooling layers, and fully connected layers, to progressively learn complex representations of image data. In this project, a CNN model was developed to classify traffic sign images from the GTSRB dataset into 43 different categories. The network learns distinctive characteristics of each traffic sign and uses this knowledge to accurately predict the correct class of unseen images. Due to their high accuracy and ability to handle complex visual data, CNNs have become one of the most widely used architectures in computer vision applications such as image recognition, object detection, medical imaging, and autonomous driving systems.


# 4.2 CNN Architecture

The architecture of a Convolutional Neural Network (CNN) defines how different layers are organized to extract features and perform image classification. In this project, the CNN architecture consists of multiple convolutional layers, pooling layers, dropout layers, a flatten layer, and fully connected dense layers. The convolutional layers are responsible for extracting important visual features such as edges, shapes, and patterns from traffic sign images. Max pooling layers reduce the spatial dimensions of feature maps, decreasing computational complexity while retaining important information. Dropout layers are included to prevent overfitting by randomly disabling a fraction of neurons during training. After feature extraction, the flatten layer converts the two-dimensional feature maps into a one-dimensional vector, which is then passed to fully connected dense layers for classification. Finally, a Softmax output layer with 43 neurons is used to predict the probability of each traffic sign class. This hierarchical architecture enables the CNN to learn increasingly complex image features and achieve accurate traffic sign classification.

# 4.3 Convolution Layer

The Convolution Layer is the fundamental building block of a Convolutional Neural Network (CNN) and is primarily responsible for feature extraction from input images. In this layer, a small matrix known as a filter or kernel slides across the image and performs mathematical operations to detect important visual patterns such as edges, corners, textures, shapes, and symbols. Each filter focuses on learning a specific feature from the image, and multiple filters are used simultaneously to capture different characteristics of traffic signs. As the image passes through successive convolutional layers, the network learns increasingly complex features, starting from simple edges and progressing to complete traffic sign patterns. In this project, convolutional layers play a crucial role in identifying distinctive features of the 43 traffic sign classes present in the GTSRB dataset. The extracted feature maps serve as the foundation for accurate classification by the subsequent layers of the CNN.


# 4.4 Activation Function (ReLU)

The Rectified Linear Unit (ReLU) is one of the most commonly used activation functions in deep learning and plays a crucial role in Convolutional Neural Networks (CNNs). After the convolution operation, the ReLU activation function is applied to introduce non-linearity into the network, enabling the model to learn complex patterns and relationships present in image data. Mathematically, ReLU converts all negative values to zero while retaining positive values unchanged. This simple operation helps the network focus on important features and improves computational efficiency. Compared to traditional activation functions such as Sigmoid and Tanh, ReLU reduces the vanishing gradient problem and accelerates the training process. In this project, ReLU activation is applied after each convolutional layer, allowing the CNN to effectively learn and distinguish various traffic sign features from the GTSRB dataset. Its ability to improve convergence speed and model performance makes it a fundamental component of modern deep learning architectures.


# 4.5 Max Pooling Layer

The Max Pooling Layer is an important component of a Convolutional Neural Network (CNN) that is used to reduce the spatial dimensions of feature maps while preserving the most important information. After the convolution and activation operations, the feature maps may contain a large amount of data, which increases computational complexity and training time. Max pooling addresses this issue by dividing the feature map into small regions and selecting the maximum value from each region. This process reduces the size of the feature maps, decreases the number of parameters, and minimizes memory usage while retaining the most significant features. Additionally, max pooling helps make the CNN more robust to small variations, noise, and positional changes in images. In this project, max pooling layers are used after convolutional layers to efficiently extract the most relevant traffic sign features and improve the overall performance of the classification model.


#4.6 Dropout Layer

The Dropout Layer is a regularization technique used in Convolutional Neural Networks (CNNs) to reduce overfitting and improve the model’s ability to generalize to unseen data. During the training process, dropout randomly deactivates a certain percentage of neurons in a layer, preventing the network from becoming overly dependent on specific neurons. As a result, the model learns more robust and diverse feature representations rather than memorizing the training data. This technique encourages the network to distribute learning across multiple neurons and reduces the risk of overfitting, especially when working with complex models and limited datasets. In this project, dropout layers were incorporated after convolutional and fully connected layers to improve the CNN’s generalization capability and achieve better traffic sign classification performance on unseen test images. By reducing overfitting, dropout contributes significantly to the stability and reliability of the trained model.



```python
#Build CNN model
model = Sequential()

model.add( Input(shape = X_train.shape[1:]))

model.add(Conv2D(32,(3,3), activation='relu'))

model.add( BatchNormalization())

model.add(Conv2D(32,(3,3),activation='relu'))

model.add( MaxPooling2D((2,2)))

model.add( Dropout(0.25))
```

```python
#Add second CNN block
model.add(Conv2D(64,(3,3),activation='relu'))

model.add( BatchNormalization())

model.add( Conv2D( 64,(3,3), activation='relu'))

model.add( MaxPooling2D((2,2)))

model.add( Dropout( 0.25))
```

```python
#Add Flatten + Dense layers
model.add( Flatten())

model.add(Dense(256,activation='relu'))

model.add(BatchNormalization())

model.add(Dropout(0.5))

model.add(Dense(43,activation='softmax'))
```

```python
model.summary()
```

4.11 Model Training

Model training is the process through which the Convolutional Neural Network (CNN) learns patterns and features from the training dataset. During training, the traffic sign images are passed through the network, and predictions are generated for each image. The predicted outputs are then compared with the actual class labels using the loss function, and the error is calculated. This error is propagated backward through the network using the backpropagation algorithm, allowing the model to update its weights and improve its predictions. In this project, the CNN was trained using the GTSRB dataset for multiple epochs, enabling the model to gradually learn the distinguishing characteristics of 43 different traffic sign classes. Techniques such as data augmentation, dropout, Early Stopping, and learning rate reduction were applied to improve model generalization and prevent overfitting. As training progressed, the model achieved higher accuracy and lower loss, indicating successful learning of traffic sign features.


```python
#Compile the model
model.compile(

    optimizer='adam',

    loss='categorical_crossentropy',

    metrics=['accuracy']

)
```

```python
#Train the CNN model
history = model.fit( X_train,y_train,epochs=15,batch_size=32,validation_data=(X_test,  y_test))
```

# 6.2 ResNet50 vs CNN

ResNet50 and Convolutional Neural Networks (CNNs) are both deep learning models used for image classification and computer vision tasks. However, ResNet50 is an advanced architecture built on the fundamental concepts of CNN. While a standard CNN consists of convolutional, pooling, and fully connected layers, ResNet50 introduces residual (skip) connections, which allow information to bypass certain layers. This innovation helps solve the vanishing gradient problem, enabling the training of much deeper networks without performance degradation.

A traditional CNN is suitable for small to medium-sized image classification problems and is relatively simple to design and train. However, as the number of layers increases, standard CNNs may experience reduced accuracy due to vanishing gradients and overfitting. In contrast, ResNet50 contains 50 layers and uses residual learning to preserve feature information throughout the network, resulting in better feature extraction and higher classification accuracy.
